```python
# A) 備份 LoRA（GGUF，217MB）到雲端硬碟，並強制上傳所有待傳檔案
import os, shutil, subprocess
from google.colab import drive
if not os.path.isdir('/content/drive/MyDrive'):
    drive.mount('/content/drive')
dst = '/content/drive/MyDrive/MeetSum_finetune/recovered'
os.makedirs(dst, exist_ok=True)
for f in ['/content/ft_export/typo-lora.gguf', '/content/ft_export/Modelfile']:
    shutil.copy(f, dst)
print(subprocess.run('ls -la /content/drive/MyDrive/MeetSum_finetune/recovered /content/drive/MyDrive/MeetSum_finetune/run1 /content/drive/MyDrive/MeetSum_finetune/run1/* | head -40', shell=True, capture_output=True, text=True).stdout)
drive.flush_and_unmount()
print('上傳完成')
```
